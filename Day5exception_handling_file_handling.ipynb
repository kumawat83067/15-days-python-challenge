# Why exception is dangerous?

- sudden termination of a program

```python
n1 = int(input("enter a number:"))
n2 = int(input("enter b number:"))

try:
    res = n1/n2
except:
    print('something went wrong')
else:
    print('Result :',res)
finally:
    print("Happy world")
```

# How to handle multiple exception 


```python
try:
    n1 = int(input("enter a number :"))
    n2 = int(input("enter b number :"))
    res = n1/n2
except (ZeroDivisionError , ValueError):
    print("Something went wrong")
else:
    print("Result:",res)
finally:
    print(("happy world"))
```

```python
try:
    n1 = int(input("enter a number :"))
    n2 = int(input("enter b number :"))
    res = n1/n2
except Exception as msg:
    print("Exception occur:",msg)
    print(msg.__class__)
    print(type(msg))
else:
    print("Result:",res)
finally:
    print("happy world")
```

# Insufficient Balance Error

```python
class LessBalanceException(Exception):
    pass

balance = 11000
withdraw = int(input("enter withdraw amount:"))
try:
    if balance - withdraw < 0:
        raise LessBalanceException("Balance is low")
    else:
        balance = balance - withdraw
        print("Successfully withdraw the amount")
except Exception as e:
    print(e);
else:
    print(balance);
finally:
    print("withdraw process is completed.")
```

# Operation in a file 
- Text
  - 'read'
    - r mode is use to read the data from a file
    - file must be exist if file is not exist it will raise an error
  - 'write'
    - w mode is use to write the data into a file
    - if file is not exist than it will create a new file else it will override the data in existing file
  - ' append'
    - a mode is use to add the data after the existing data inside the file
    - it will not override the data
    - if file is not exist it will create a file
  - 'read and write'
    - r+
    - r+ mode is use to read and write the data into a file
    - file must be exist
    - if file is exist and we are writting the data than it will add the  data into a file not override the existing data
  - 'write and read'
    - w+ mode
    - w+ mode is use to write and read the data from a file
    - over ride  the data if file is exist

```python
f = open('attendance.csv','a')
num = int(input('enter a number of student:->'))
for i in range (1, num+1):
    name = input(f"Enter a name{i}:")
    date = date.today()
    course = input('enter course:')
    status = input('enter status P/A:').upper()
    if status == "A" or status =="P":
        data = f"\n{name},{date},{course},{status}"
        f.write(data)
    else:
        print("give correct status")
        break
else:
    print("data is inserted successfully")
    f.close()
```
