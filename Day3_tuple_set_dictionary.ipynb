# Tuple Vs List

# tuple

- immutable data type
- ()
- class tuple
- tuple take less memory
- tuple is fast
- tuple is use to store constant data
- it can be hashable 

# list

- []
- mutable data type
- class list
- list take more memory
- list is slow
- list is ust to store dynamic data
- can not be hashabale

# Time

```python
import timeit
```

```python
time_lis = timeit.timeit(stmt = '[1,2,3,4,5]')
time_t = timeit.timeit(stmt = '(1,2,3,4,5)')
```

```python
print(f"time taken by list is :{round(time_lis,3)}")
print("time taken by tuple is :",round(time_t,3))
```

- values = return all the values present in our dictionary(_.value())
- keys = return all the keys present in the dictionary (_.keys())

- fromkeys = use to create a new dic. from a sequence of keys , with the same default value for each key.(_fromkeys(' '))
- pop = use to remove a specified key from a dictionary and  return its value (_.pop(' '))
- popitem = removes and returns the last inserted key-value pair from a dic.(_.popitme())
- setdefault = return the value of a key . if the key does not exist , it adds the key with a default value.(_.setdefault(key,default_value))
- update = use to add new key - value or modify existing values in a dictionary
