Question: Write a Python program to find the largest among three numbers.

```python
a = int(input("enter first number:"))
b = int(input("enter second number:"))
c = int(input("enter third number:"))

if  a >= b and  a >= c :
    print("Largest:",a)

elif b >= a and b>=c :
    print("Largest:",b)

else:
    print("Largest:",c)
```

Question: Write a Python program to find the factorial of a number.

```python
num = int(input("enter a number:"))
fact = 1
for i  in range(1,num+1):
    fact = fact * i 
print("factorial:",fact)
```

Question: Write a Python program to find the sum of numbers from 1 to N.

```python
n = int(input("enter  a number:"))
total  = 0 
for  i in range(1 , n+1):
    total = total + i
print("Sum:",total)
```

Question: Write a Python program to reverse a number.

```python
num = int(input("enter a  number:"))
rev = 0
while num>0:
    digit  = num % 10 
    rev = rev  * 10 + digit 
    num = num // 10 
print("reverse:", rev)
```

Question: Write a Python program to print the first N Fibonacci numbers.

```python
n = int(input("enter  a number:"))
a  = 0 
b = 1
for i  in range(n):
    print(a , end=" ")
    c = a+b
    a = b
    b = c
```
