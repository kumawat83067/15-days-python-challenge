- upper = convert a string to upper case
- lower = convert a string to lower case
- capitalize = convert  first letter of string to capital letter  and completer string to lower case
- title = convert a string to title case string (in string each word's first letter convert to capital letter)
- casefold = it is used to convert a string into a normalized lowercase form mainly for case - insensitive string comparisons.
- difference in casefold and  lower case is    lower case is basic and casefold is for case-insensitive comparison
- swapcase = conver the upper case to lower case and vice versa
- count = return the count of char or substring from start to end of the string
- replace = replace method is used to replace a part of a string with another string
- index = it is used to find the position of a specified element or substring(value error when not found)
- rindex = it from right to left
- find = it is used to find the position of a substirng inside a string (no error found when string not found)
- rfind = find right to left
- split = is used to divide a string into a list of smaller based on a separator
- rsplit = split from right side
- partition = it will divide the string into three parts
- rpartition = it will divide from right side
- strip = it is used to remove unwanted characters from the beginning and end of the string
- lstrip = remove fromm left only
- rstrip = remove from right only
- center = it is used to center align a string by adding padding character on both sides
- ljust = it is used to left align a string by adding padding characters on the right side .
- rjust = it is used to right align a string by adding padding character on left side
- zfill = it is used to add zero to the left side of a string until it reach the specified width
- encode = it is used to convert a string into bytes using a specified encoding , usually UTF-8
- join = it is used to combine mutliple string into a single string using a specified separator.

- isalnum = if the string is alpha numeric string than return true otherwise return false
- isalpha = if the string is alphabtic string tha return true other wise return false
- isascii = if the string contain ascii string than return true otherwise return false
- isdecimal = if the string only decimal value than return true other wise false
- isdigit = if the string contain only digit than return true otherwise return false
- isnumeric = if the string is contain numeric value than return true otherwise return false
- isidentifier = if the string is identifier than return true otherwise return false
- istitle = if the string is in title case than return true otherwise return false
- isupper = if string is in upper case than return true otherwise return false
- islower = if string is in lower case than return true otherwise return false
- isprintable = if the string is printable than return true otherwise return false
- isspace = if the string contain only whitespace than return true otherwise return false
