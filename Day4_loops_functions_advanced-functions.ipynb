# Nested Loop

```python
i = 1
while i <= 5:
    j = 1
    while j<=5:
        print(f"the value of i is {i} and teh value of j is{j}")
        j+=1
    print("*"*50)
    i+=1
```

```python
i = 1
while i<=5:
    j = 1
    while  j<=5:
        print("*", end=" ")
        j+=1
    print()
    i+=1
```

```python
i = 1
while i<=5:
    j = 1
    while  j<=5:
        if j<=i:
            print("*", end=" ")
        j+=1
    print()
    i+=1
```

# Scope

- scope is a area where a variable can be accessible

# Two type of scope

- global scope
- local scope

- Global scope = it is our main space (main memory)
- Local scope = local scope is a area which is created after calling the function or after calling the class

Global variable = it is define outside the local space (variable define inside the main space)
- note :  global variable can be accessible value inside the class or function

Local variable = if is define inside the class or function we can not access the local variable outside the function of class

```python
def fun():
    a = 11 # local variable 
    print("The value of a inside the function is :",a)
```
